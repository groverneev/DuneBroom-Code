```python
# Cell 2: Imports & check GPU status
from ultralytics import YOLO
!nvidia-smi
```

```python

# Define dataset path and check data.yaml
import os
dataset_path = "C:/Users/grove/Documents/SRP_24-25/final_roboflow_data"
data_yaml = f"{dataset_path}/data.yaml"

assert os.path.exists(data_yaml), f"{data_yaml} not found! Please check your dataset setup."
print(f"Using dataset config: {data_yaml}")
```

```python
# Load model and start training
model = YOLO("C:/Users/grove/Downloads/yolo11m.pt")     # loads pretrained yolov11 small model

results = model.train(
    data=data_yaml,
    epochs=100,
    imgsz=640,
    batch=16,
    project='runs',
    name='yolov11_custom',
    dfl=False
)
```

```python
# Validate the model on the validation dataset
metrics = model.val(data=data_yaml)
print(metrics)
```

```python
# Export trained model to ONNX
model.export(format='onnx')
```
